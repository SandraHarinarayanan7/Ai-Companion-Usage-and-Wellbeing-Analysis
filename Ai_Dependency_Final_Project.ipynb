```python
data=data.rename(columns={
    'user_id':'User_ID',
    'age':'Age',
    'gender':'Gender',
    'occupation_status':'Occupation_Status',
    'education_level':'Education_Level',
    'income_level':'Income_Level',
    'relationship_status':'Relationship_Status',
    'living_situation':'Living_Situation',
    'country_region':'Country_Region',
    'ai_chatbot_platform':'Ai_Chatbot_Platform',
    'primary_use_case':'Primary_Use_Case',
    'daily_ai_chat_hours':'Daily_Ai_Chat_Hours',
     'human_social_interaction_hours':'Human_Social_Interaction_Hours',
    'social_media_hours_daily':'Social_Media_Hours_Daily',
     'screen_time_total_hours':'Screen_Time_Total_Hours',
     'number_of_close_friends':'Number_Of_Close_Friends',
    'loneliness_score':'Loneliness_Score',
    'emotional_attachment_score':'Emotional_Attachment_Score',
    'real_relationship_satisfaction':'Real_Relationship_Satisfaction',
    'prefers_ai_over_humans':'Prefers_Ai_Over_Humans',
    'sleep_hours':'Sleep_Hours',
    'sleep_quality_score':'Sleep_Quality_Score',
    'exercise_hours_weekly':'Exercise_Hours_Weekly',
    'anxiety_score':'Anxiety_Score',
    'depression_score':'Depression_Score',
    'stress_score':'Stress_Score',
    'self_esteem_score':'Self_Esteem_Score',
    'academic_or_work_performance_change':'Academic_Or_Work_Performance_Change',
     'therapy_attendance':'Therapy_Attendance',
     'dependency_risk_label':'Dependency_Risk_Label'
})
data
```

```python
def Stress_Pattern_Category(Stress_Score):
    if Stress_Score <= 3:
        return "Calm"
    elif Stress_Score <= 5:
        return "Elevated"
    elif Stress_Score <= 7:
        return "Strained"
    else:
        return "High-Strain"

data["Stress_Pattern"] = data["Stress_Score"].apply(Stress_Pattern_Category)

data
```

```python
def Age_Group(Age):
    if Age < 20:
        return "15-19"
    elif Age < 25:
        return "20-24"
    elif Age < 35:
        return "25-34"
    else:
        return "35-44"

data["Age_Group"] = data["Age"].apply(Age_Group)

data
```

```python
import matplotlib.pyplot as plt
use_case_count=data['Primary_Use_Case'].value_counts()
plt.figure(figsize=(6,6))
plt.pie(use_case_count,labels=use_case_count.index,autopct='%1.1f%%')
plt.title("Primary AI Use Cases Distribution")
plt.show()
```

```python
age_data = data.groupby(['Age_Group', 'Dependency_Risk_Label'])['User_ID'].count().unstack()

age_data.plot(kind='bar', stacked=True, figsize=(8,6))

plt.title("Dependency Risk Levels Across Age Groups")
plt.xlabel("Age Group")
plt.ylabel("Count of User ID")
plt.xticks(rotation=0)
plt.legend(title="Dependency Risk Label")
plt.show()
```

```python
age_data = data.groupby('Age_Group')[['Exercise_Hours_Weekly','Social_Media_Hours_Daily']].mean()

age_data.plot(kind='line', figsize=(8,6), marker='o')

plt.title("Exercise & Social Media Usage by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Avg. Hours")
plt.xticks(rotation=0)
plt.legend(["Avg. Exercise Hours", "Avg. Social Media Hours"])
plt.show()
```

```python
risk_count = data['Dependency_Risk_Label'].value_counts()

plt.figure(figsize=(7,7))

plt.pie(risk_count, labels=risk_count.index, autopct='%1.1f%%', startangle=90, wedgeprops=dict(width=0.4))

plt.title("Dependency Risk Level Distribution")

plt.show()
```

```python
plt.figure(figsize=(8,6))

plt.scatter(data['Daily_Ai_Chat_Hours'], data['Loneliness_Score'])

plt.title("AI Chat Usage vs. Loneliness Score")
plt.xlabel("Average Daily AI Chat Hours")
plt.ylabel("Average Loneliness Score")

plt.show()
```

```python
import seaborn as sns
stress_data = data.groupby('Stress_Pattern')[['Emotional_Attachment_Score', 'Loneliness_Score']].mean()

plt.figure(figsize=(8,5))
sns.heatmap(stress_data, annot=True, fmt='.3f')

plt.title('Loneliness & Emotional Attachment By Stress Pattern')
plt.xlabel('Measures')
plt.ylabel('Stress Pattern')
plt.show()
```

```python
platform_data = data.groupby('Ai_Chatbot_Platform')[['Daily_Ai_Chat_Hours', 'Screen_Time_Total_Hours']].mean()

platform_data.plot(kind='bar', stacked=True, figsize=(8,5))

plt.title('Average AI Chat Hours and Screen Time by AI Chatbot Platform')
plt.xlabel('AI Chatbot Platform')
plt.ylabel('Average Hours')
plt.xticks(rotation=0)
plt.show()
```
